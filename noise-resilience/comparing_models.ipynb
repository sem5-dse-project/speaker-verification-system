```python
import os
import torch
import torchaudio
import pandas as pd
import torch.nn.functional as F
import random
import numpy as np
from transformers import Wav2Vec2FeatureExtractor, Wav2Vec2Model

try:
    from speechbrain.inference.speaker import EncoderClassifier
except ImportError:
    from speechbrain.pretrained import EncoderClassifier

os.environ["HF_HUB_OFFLINE"] = "1"

# ==========================================
# 1. DEVICE DETECTION (Targets GPU if available)
# ==========================================
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"🚀 Running on device: {device.type.upper()}")

def calculate_eer(pos_scores, neg_scores):
    if len(pos_scores) == 0 or len(neg_scores) == 0:
        return 0.0, 0.0

    y_scores = np.array(pos_scores + neg_scores)
    thresholds = np.sort(y_scores)

    min_diff = float('inf')
    best_eer = 1.0
    best_thresh = 0.0

    for t in thresholds:
        far = np.mean(np.array(neg_scores) >= t)
        frr = np.mean(np.array(pos_scores) < t)

        diff = abs(far - frr)
        if diff < min_diff:
            min_diff = diff
            best_eer = (far + frr) / 2.0
            best_thresh = t

    return best_eer * 100, best_thresh

def run_sinhala_comparison():
    print("1. Parsing Local Sinhala Dataset...")

    # Load metadata
    metadata = pd.read_csv(
        'utt_spk_text.tsv',
        sep='\t',
        header=None,
        names=['FileID', 'UserID', 'Transcription']
    )

    # Get local files from your custom function
    local_files, paths = get_files()
    available_data = metadata[metadata['FileID'].isin(local_files)]

    # Group available files by speaker
    spk_to_files = available_data.groupby('UserID')['FileID'].apply(list).to_dict()
    all_speakers = list(spk_to_files.keys())

    # 2. Construct Balanced Evaluation Trials
    print("2. Generating Evaluation Trials...")
    positive_pairs = []
    negative_pairs = []

    MAX_PAIRS = 5000
    upper_bound = 700_000

    for spk, files in spk_to_files.items():
        if len(files) >= 2:
            for i in range(len(files)):
                for j in range(i + 1, len(files)):
                    positive_pairs.append((files[i], files[j]))
                    if len(positive_pairs) % 50_000 == 0:
                        print(f"   -> Generated {len(positive_pairs)} Same-Speaker trials.")

    random.seed(42)
    random.shuffle(positive_pairs)
    positive_pairs = positive_pairs[:MAX_PAIRS]

    for f1, _ in positive_pairs:
        spk1 = available_data[available_data['FileID'] == f1]['UserID'].values[0]
        other_speakers = [s for s in all_speakers if s != spk1]

        if other_speakers:
            random_other_spk = random.choice(other_speakers)
            random_other_file = random.choice(spk_to_files[random_other_spk])
            negative_pairs.append((f1, random_other_file))

            if len(negative_pairs) % 50_000 == 0:
                print(f"   -> Generated {len(negative_pairs)} Different-Speaker trials.")

    random.shuffle(negative_pairs)

    negative_pairs = negative_pairs[:len(positive_pairs)]

    print(f"   -> Generated {len(positive_pairs)} Same-Speaker trials.")
    print(f"   -> Generated {len(negative_pairs)} Different-Speaker trials.")

    eval_files = list(set([f for pair in positive_pairs for f in pair] +
                          [f for pair in negative_pairs for f in pair]))
    print(f"   -> Found {len(eval_files)} unique files to process.")

    # We keep torchaudio loader on CPU (standard practice as file I/O is handled on CPU)
    def load_and_resample(file_id):
        path = paths[file_id]
        waveform, sr = torchaudio.load(path)
        if sr != 16000:
            resampler = torchaudio.transforms.Resample(orig_freq=sr, new_freq=16000)
            waveform = resampler(waveform)
        if waveform.shape[0] > 1:
            waveform = torch.mean(waveform, dim=0, keepdim=True)
        return waveform.squeeze().numpy()

    # ==========================================
    # 3. LOAD MODELS DIRECTLY TO GPU
    # ==========================================
    print("\n3. Loading Models onto GPU...")
    # Speechbrain uses 'run_opts' dictionary to automatically set target device
    ecapa_classifier = EncoderClassifier.from_hparams(
        source="tmpdir",
        savedir="tmpdir",
        run_opts={"device": str(device)}
    )

    xlsr_extractor = Wav2Vec2FeatureExtractor.from_pretrained("./xlsr_model_dir")

    # Hugging Face models are pushed to GPU explicitly via `.to(device)`
    xlsr_model = Wav2Vec2Model.from_pretrained("./xlsr_model_dir").to(device)
    xlsr_model.eval()

    # ==========================================
    # 4. EXPLICITLY PUSH INPUTS TO GPU
    # ==========================================
    def get_ecapa_embedding(audio_array):
        signal = torch.tensor(audio_array).unsqueeze(0).float().to(device)
        with torch.no_grad():
            emb = ecapa_classifier.encode_batch(signal).squeeze(1)
            return F.normalize(emb, p=2, dim=-1)

    def get_xlsr_embedding(audio_array):
        inputs = xlsr_extractor(audio_array, sampling_rate=16000, return_tensors="pt", padding=True)
        # Hugging Face inputs (dictionary of tensors) need to be recursively pushed to GPU
        inputs = {k: v.to(device) for k, v in inputs.items()}
        with torch.no_grad():
            outputs = xlsr_model(**inputs)
            emb = outputs.last_hidden_state.mean(dim=1)
            return F.normalize(emb, p=2, dim=-1)

    # 5. Single-Pass Embedding Extraction
    print("\n4. Running Single-Pass Inference on local Sinhala files...")
    ecapa_cache = {}
    xlsr_cache = {}

    for idx, f_id in enumerate(eval_files):
        if idx % 50 == 0 or idx == len(eval_files) - 1:
            print(f"   -> Processing file {idx + 1}/{len(eval_files)}")

        try:
            audio_data = load_and_resample(f_id)
            ecapa_cache[f_id] = get_ecapa_embedding(audio_data)
            xlsr_cache[f_id] = get_xlsr_embedding(audio_data)
        except Exception as e:
            print(f"      [Warning] Error loading {f_id}: {e}")
            continue

    # 6. Evaluate Pairs using cached GPU embeddings
    print("\n5. Computing Pairwise Similarity Scores...")
    cos = torch.nn.CosineSimilarity(dim=-1)

    ecapa_pos_scores = []
    ecapa_neg_scores = []
    xlsr_pos_scores = []
    xlsr_neg_scores = []

    for f1, f2 in positive_pairs:
        if f1 in ecapa_cache and f2 in ecapa_cache:
            # .item() pulls the single-element GPU tensor back to a CPU float
            ecapa_pos_scores.append(cos(ecapa_cache[f1], ecapa_cache[f2]).item())
            xlsr_pos_scores.append(cos(xlsr_cache[f1], xlsr_cache[f2]).item())

    for f1, f2 in negative_pairs:
        if f1 in ecapa_cache and f2 in ecapa_cache:
            ecapa_neg_scores.append(cos(ecapa_cache[f1], ecapa_cache[f2]).item())
            xlsr_neg_scores.append(cos(xlsr_cache[f1], xlsr_cache[f2]).item())

    # 7. Calculate Metrics
    ecapa_avg_pos = np.mean(ecapa_pos_scores) if ecapa_pos_scores else 0
    ecapa_avg_neg = np.mean(ecapa_neg_scores) if ecapa_neg_scores else 0
    ecapa_margin = ecapa_avg_pos - ecapa_avg_neg
    ecapa_eer, ecapa_thresh = calculate_eer(ecapa_pos_scores, ecapa_neg_scores)

    xlsr_avg_pos = np.mean(xlsr_pos_scores) if xlsr_pos_scores else 0
    xlsr_avg_neg = np.mean(xlsr_neg_scores) if xlsr_neg_scores else 0
    xlsr_margin = xlsr_avg_pos - xlsr_avg_neg
    xlsr_eer, xlsr_thresh = calculate_eer(xlsr_pos_scores, xlsr_neg_scores)

    # 8. Print Results Table
    print("\n" + "="*70)
    print("RESULTS: MASS EVALUATION ON SINHALA AUDIO DATASET")
    print("="*70)

    metrics_summary = pd.DataFrame({
        "Metric": [
            "Average Same-Speaker Sim",
            "Average Diff-Speaker Sim",
            "Separation Margin (↑)",
            "Best Decision Threshold",
            "Equal Error Rate (EER % ↓)"
        ],
        "ECAPA-TDNN": [
            f"{ecapa_avg_pos:.4f}",
            f"{ecapa_avg_neg:.4f}",
            f"{ecapa_margin:.4f}",
            f"{ecapa_thresh:.4f}",
            f"{ecapa_eer:.2f}%"
        ],
        "XLS-R 300M": [
            f"{xlsr_avg_pos:.4f}",
            f"{xlsr_avg_neg:.4f}",
            f"{xlsr_margin:.4f}",
            f"{xlsr_thresh:.4f}",
            f"{xlsr_eer:.2f}%"
        ]
    })
    print(metrics_summary.to_string(index=False))
    print("="*70)

run_sinhala_comparison()
```

```python
import os
import torch
import torchaudio
import pandas as pd
import torch.nn.functional as F
import random
import numpy as np
from transformers import Wav2Vec2FeatureExtractor, Wav2Vec2Model

try:
    from speechbrain.inference.speaker import EncoderClassifier
except ImportError:
    from speechbrain.pretrained import EncoderClassifier

# Set to "0" if you want to pull models automatically from Hugging Face / SpeechBrain
os.environ["HF_HUB_OFFLINE"] = "1"

# ==========================================
# 1. DEVICE DETECTION (Targets GPU if available)
# ==========================================
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"🚀 Running on device: {device.type.upper()}")

def calculate_eer(pos_scores, neg_scores):
    if len(pos_scores) == 0 or len(neg_scores) == 0:
        return 0.0, 0.0

    y_scores = np.array(pos_scores + neg_scores)
    thresholds = np.sort(y_scores)

    min_diff = float('inf')
    best_eer = 1.0
    best_thresh = 0.0

    for t in thresholds:
        far = np.mean(np.array(neg_scores) >= t)
        frr = np.mean(np.array(pos_scores) < t)

        diff = abs(far - frr)
        if diff < min_diff:
            min_diff = diff
            best_eer = (far + frr) / 2.0
            best_thresh = t

    return best_eer * 100, best_thresh

def run_librispeech_comparison():
    # ==========================================
    # 2. LOAD LIBRISPEECH DEV-CLEAN DATASET
    # ==========================================
    print("1. Loading LibriSpeech dev-clean dataset...")
    # Downloads ~337 MB of dev-clean into ./ if not already present
    dataset = torchaudio.datasets.LIBRISPEECH(root="./", url="dev-clean", download=True)
    print(f"   -> Loaded {len(dataset)} total utterances.")

    # Speed up parsing: Build an index mapping of speaker IDs using fast metadata inspection
    print("   -> Mapping speaker identities...")
    idx_to_spk = {}
    for idx in range(len(dataset)):
        # get_metadata yields speaker ID without reading the heavy raw FLAC waveforms into RAM
        _, _, _, speaker_id, _, _ = dataset.get_metadata(idx)
        idx_to_spk[idx] = speaker_id

    # Group dataset indices by speaker
    spk_to_indices = {}
    for idx, spk in idx_to_spk.items():
        if spk not in spk_to_indices:
            spk_to_indices[spk] = []
        spk_to_indices[spk].append(idx)

    all_speakers = list(spk_to_indices.keys())
    print(f"   -> Found {len(all_speakers)} unique speakers in the split.")

    # ==========================================
    # 3. GENERATE TRIAL PAIRS
    # ==========================================
    print("2. Generating Evaluation Trials...")
    positive_pairs = []
    negative_pairs = []

    MAX_PAIRS = 5000  # Number of verification trials to test (adjustable)

    # Generate same-speaker positive trials
    for spk, indices in spk_to_indices.items():
        if len(indices) >= 2:
            for i in range(len(indices)):
                for j in range(i + 1, len(indices)):
                    positive_pairs.append((indices[i], indices[j]))

    random.seed(42)
    random.shuffle(positive_pairs)
    positive_pairs = positive_pairs[:MAX_PAIRS]

    # Generate balanced different-speaker negative trials
    for idx1, _ in positive_pairs:
        spk1 = idx_to_spk[idx1]
        other_speakers = [s for s in all_speakers if s != spk1]

        if other_speakers:
            random_other_spk = random.choice(other_speakers)
            random_other_idx = random.choice(spk_to_indices[random_other_spk])
            negative_pairs.append((idx1, random_other_idx))

    negative_pairs = negative_pairs[:len(positive_pairs)]

    print(f"   -> Generated {len(positive_pairs)} Same-Speaker trials.")
    print(f"   -> Generated {len(negative_pairs)} Different-Speaker trials.")

    # Gather all unique indices we actually need to compute embeddings for
    eval_indices = list(set([idx for pair in positive_pairs for idx in pair] +
                            [idx for pair in negative_pairs for idx in pair]))
    print(f"   -> Found {len(eval_indices)} unique utterances to process.")

    # Audio loading function tailored for LibriSpeech output
    def load_and_resample(dataset_idx):
        waveform, sr, _, _, _, _ = dataset[dataset_idx]
        if sr != 16000:
            resampler = torchaudio.transforms.Resample(orig_freq=sr, new_freq=16000)
            waveform = resampler(waveform)
        if waveform.shape[0] > 1:
            waveform = torch.mean(waveform, dim=0, keepdim=True)
        return waveform.squeeze().numpy()

    # ==========================================
    # 4. LOAD MODEL EMBEDDING ENGINES
    # ==========================================
    print("\n3. Loading Models onto GPU...")

    # --- MODEL 1: ECAPA-TDNN ---
    # Offline source: "tmpdir" / Online source: "speechbrain/spkrec-ecapa-voxceleb"
    ecapa_source = "tmpdir" if os.environ["HF_HUB_OFFLINE"] == "1" else "speechbrain/spkrec-ecapa-voxceleb"
    ecapa_classifier = EncoderClassifier.from_hparams(
        source=ecapa_source,
        savedir="tmpdir",
        run_opts={"device": str(device)}
    )

    # --- MODEL 2: XLS-R ---
    # Offline source: "./xlsr_model_dir" / Online source: "facebook/wav2vec2-xls-r-300m"
    xlsr_source = "./xlsr_model_dir" if os.environ["HF_HUB_OFFLINE"] == "1" else "facebook/wav2vec2-xls-r-300m"
    xlsr_extractor = Wav2Vec2FeatureExtractor.from_pretrained(xlsr_source)
    xlsr_model = Wav2Vec2Model.from_pretrained(xlsr_source).to(device)
    xlsr_model.eval()

    def get_ecapa_embedding(audio_array):
        signal = torch.tensor(audio_array).unsqueeze(0).float().to(device)
        with torch.no_grad():
            emb = ecapa_classifier.encode_batch(signal).squeeze(1)
            return F.normalize(emb, p=2, dim=-1)

    def get_xlsr_embedding(audio_array):
        inputs = xlsr_extractor(audio_array, sampling_rate=16000, return_tensors="pt", padding=True)
        inputs = {k: v.to(device) for k, v in inputs.items()}
        with torch.no_grad():
            outputs = xlsr_model(**inputs)
            emb = outputs.last_hidden_state.mean(dim=1)
            return F.normalize(emb, p=2, dim=-1)

    # ==========================================
    # 5. SINGLE-PASS CACHED INFERENCE
    # ==========================================
    print("\n4. Running Single-Pass Inference on LibriSpeech dev-clean files...")
    ecapa_cache = {}
    xlsr_cache = {}

    for idx, eval_idx in enumerate(eval_indices):
        if idx % 50 == 0 or idx == len(eval_indices) - 1:
            print(f"   -> Processing file {idx + 1}/{len(eval_indices)}")

        try:
            audio_data = load_and_resample(eval_idx)
            ecapa_cache[eval_idx] = get_ecapa_embedding(audio_data)
            xlsr_cache[eval_idx] = get_xlsr_embedding(audio_data)
        except Exception as e:
            print(f"      [Warning] Error processing index {eval_idx}: {e}")
            continue

    # ==========================================
    # 6. EVALUATE EMBEDDINGS
    # ==========================================
    print("\n5. Computing Pairwise Similarity Scores...")
    cos = torch.nn.CosineSimilarity(dim=-1)

    ecapa_pos_scores = []
    ecapa_neg_scores = []
    xlsr_pos_scores = []
    xlsr_neg_scores = []

    for idx1, idx2 in positive_pairs:
        if idx1 in ecapa_cache and idx2 in ecapa_cache:
            ecapa_pos_scores.append(cos(ecapa_cache[idx1], ecapa_cache[idx2]).item())
            xlsr_pos_scores.append(cos(xlsr_cache[idx1], xlsr_cache[idx2]).item())

    for idx1, idx2 in negative_pairs:
        if idx1 in ecapa_cache and idx2 in ecapa_cache:
            ecapa_neg_scores.append(cos(ecapa_cache[idx1], ecapa_cache[idx2]).item())
            xlsr_neg_scores.append(cos(xlsr_cache[idx1], xlsr_cache[idx2]).item())

    # Calculate metrics
    ecapa_avg_pos = np.mean(ecapa_pos_scores) if ecapa_pos_scores else 0
    ecapa_avg_neg = np.mean(ecapa_neg_scores) if ecapa_neg_scores else 0
    ecapa_margin = ecapa_avg_pos - ecapa_avg_neg
    ecapa_eer, ecapa_thresh = calculate_eer(ecapa_pos_scores, ecapa_neg_scores)

    xlsr_avg_pos = np.mean(xlsr_pos_scores) if xlsr_pos_scores else 0
    xlsr_avg_neg = np.mean(xlsr_neg_scores) if xlsr_neg_scores else 0
    xlsr_margin = xlsr_avg_pos - xlsr_avg_neg
    xlsr_eer, xlsr_thresh = calculate_eer(xlsr_pos_scores, xlsr_neg_scores)

    # Print results
    print("\n" + "="*70)
    print("RESULTS: MASS EVALUATION ON LIBRISPEECH (DEV-CLEAN)")
    print("="*70)

    metrics_summary = pd.DataFrame({
        "Metric": [
            "Average Same-Speaker Sim",
            "Average Diff-Speaker Sim",
            "Separation Margin (↑)",
            "Best Decision Threshold",
            "Equal Error Rate (EER % ↓)"
        ],
        "ECAPA-TDNN": [
            f"{ecapa_avg_pos:.4f}",
            f"{ecapa_avg_neg:.4f}",
            f"{ecapa_margin:.4f}",
            f"{ecapa_thresh:.4f}",
            f"{ecapa_eer:.2f}%"
        ],
        "XLS-R 300M": [
            f"{xlsr_avg_pos:.4f}",
            f"{xlsr_avg_neg:.4f}",
            f"{xlsr_margin:.4f}",
            f"{xlsr_thresh:.4f}",
            f"{xlsr_eer:.2f}%"
        ]
    })
    print(metrics_summary.to_string(index=False))
    print("="*70)

run_librispeech_comparison()
```

```python
"""
Scaled-up ECAPA-TDNN vs X-Vector comparison on Sinhala + English.

FIX SUMMARY vs. previous version:
- Uses ALL available Sinhala speakers (or a configurable target) instead of
  an arbitrary max_pairs cap that under-uses speaker diversity.
- Utterances-per-speaker is capped (not total pairs) so pair count grows
  naturally and predictably with speaker count.
- Per-model cache membership checks are already correct (kept as-is), but
  now also asserts embedding extraction succeeded for BOTH models before
  including a file in trials, avoiding silent partial-cache mismatches.
- get_files() is now an explicit required argument (a function you provide)
  instead of a bare undefined call -- fails loudly with a clear message if
  missing, instead of a cryptic NameError mid-run.

You must still provide `get_files()` yourself: it should return
(local_file_ids, paths_dict) exactly as in your original notebook.
"""

import os
import random
import itertools
import numpy as np
import pandas as pd
import torch
import torch.nn.functional as F
import torchaudio

try:
    from speechbrain.inference.speaker import EncoderClassifier
except ImportError:
    from speechbrain.pretrained import EncoderClassifier

# =====================================================================
# CONFIGURATION
# =====================================================================
OFFLINE_MODE = True

if OFFLINE_MODE:
    os.environ["HF_HUB_OFFLINE"] = "1"
    ECAPA_SOURCE = "/content/tmpdir"
    XVECT_SOURCE = "/content/x-vector"
else:
    os.environ["HF_HUB_OFFLINE"] = "0"
    ECAPA_SOURCE = "speechbrain/spkrec-ecapa-voxceleb"
    XVECT_SOURCE = "speechbrain/spkrec-xvect-voxceleb"

# --- Speaker/trial scale settings ---
NUM_SPEAKERS_SINHALA = None   # None = use ALL available speakers; or set an int, e.g. 200
UTTS_PER_SPEAKER = 4          # cap per speaker -- controls pair count growth
RANDOM_SEED = 42

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Running on device: {device.type.upper()}")


# =====================================================================
# METRICS
# =====================================================================
def calculate_eer(pos_scores, neg_scores):
    if len(pos_scores) == 0 or len(neg_scores) == 0:
        return 0.0, 0.0
    pos = np.array(pos_scores)
    neg = np.array(neg_scores)
    thresholds = np.sort(np.concatenate([pos, neg]))

    min_diff = float("inf")
    best_eer, best_thresh = 1.0, 0.0
    for t in thresholds:
        far = np.mean(neg >= t)
        frr = np.mean(pos < t)
        diff = abs(far - frr)
        if diff < min_diff:
            min_diff = diff
            best_eer = (far + frr) / 2.0
            best_thresh = t
    return best_eer * 100, best_thresh


def get_embedding(audio_array, classifier):
    signal = torch.tensor(audio_array).unsqueeze(0).float().to(device)
    with torch.no_grad():
        emb = classifier.encode_batch(signal).squeeze(1)
        return F.normalize(emb, p=2, dim=-1)


# =====================================================================
# TRIAL GENERATION -- speaker-count-driven, not pair-count-driven
# =====================================================================
def build_balanced_trials(spk_to_files: dict, num_speakers: int | None,
                           utts_per_speaker: int, seed: int = 42):
    """
    Selects up to `num_speakers` speakers (all of them if None), caps each
    speaker to `utts_per_speaker` utterances, then builds:
      - positive pairs: all within-speaker combinations of the capped utterances
      - negative pairs: one random cross-speaker match per positive pair
    This maximizes speaker diversity per trial rather than just cranking up
    a raw trial-count cap on a small speaker pool.
    """
    rng = random.Random(seed)

    eligible_speakers = [s for s, files in spk_to_files.items() if len(files) >= 2]
    rng.shuffle(eligible_speakers)
    if num_speakers is not None:
        eligible_speakers = eligible_speakers[:num_speakers]

    capped = {
        spk: spk_to_files[spk][:utts_per_speaker]
        for spk in eligible_speakers
    }

    positive_pairs = []
    for spk, files in capped.items():
        for f1, f2 in itertools.combinations(files, 2):
            positive_pairs.append((f1, f2))
    rng.shuffle(positive_pairs)

    negative_pairs = []
    speakers = list(capped.keys())
    for f1, _ in positive_pairs:
        spk1 = next(s for s, files in capped.items() if f1 in files)
        other_speakers = [s for s in speakers if s != spk1]
        other_spk = rng.choice(other_speakers)
        other_file = rng.choice(capped[other_spk])
        negative_pairs.append((f1, other_file))

    eval_keys = list({f for pair in positive_pairs for f in pair} |
                     {f for pair in negative_pairs for f in pair})

    print(f"   -> {len(capped)} speakers used, {len(eval_keys)} unique files, "
          f"{len(positive_pairs)} positive / {len(negative_pairs)} negative pairs")

    return positive_pairs, negative_pairs, eval_keys


# =====================================================================
# DATASET LOADERS
# =====================================================================
def get_sinhala_trials(get_files_fn, num_speakers, utts_per_speaker, seed):
    print("-> Parsing local Sinhala dataset...")
    metadata = pd.read_csv(
        "utt_spk_text.tsv", sep="\t", header=None,
        names=["FileID", "UserID", "Transcription"],
    )

    local_files, paths = get_files_fn()
    available = metadata[metadata["FileID"].isin(local_files)]
    spk_to_files = available.groupby("UserID")["FileID"].apply(list).to_dict()

    print(f"   -> {len(spk_to_files)} total speakers available locally "
          f"(dataset has 478 speakers overall -- only locally downloaded ones are usable here)")

    positive_pairs, negative_pairs, eval_keys = build_balanced_trials(
        spk_to_files, num_speakers, utts_per_speaker, seed
    )

    def load_fn(f_id):
        path = paths[f_id]
        waveform, sr = torchaudio.load(path)
        if sr != 16000:
            waveform = torchaudio.transforms.Resample(sr, 16000)(waveform)
        if waveform.shape[0] > 1:
            waveform = torch.mean(waveform, dim=0, keepdim=True)
        return waveform.squeeze().numpy()

    return positive_pairs, negative_pairs, eval_keys, load_fn


def get_librispeech_trials(num_speakers, utts_per_speaker, seed):
    print("-> Loading LibriSpeech dev-clean dataset...")
    dataset = torchaudio.datasets.LIBRISPEECH(root="./", url="train-clean-100", download=True)

    idx_to_spk = {}
    for idx in range(len(dataset)):
        _, _, _, speaker_id, _, _ = dataset.get_metadata(idx)
        idx_to_spk[idx] = speaker_id

    spk_to_indices = {}
    for idx, spk in idx_to_spk.items():
        spk_to_indices.setdefault(spk, []).append(idx)

    positive_pairs, negative_pairs, eval_keys = build_balanced_trials(
        spk_to_indices, num_speakers, utts_per_speaker, seed
    )

    def load_fn(dataset_idx):
        waveform, sr, _, _, _, _ = dataset[dataset_idx]
        if sr != 16000:
            waveform = torchaudio.transforms.Resample(sr, 16000)(waveform)
        if waveform.shape[0] > 1:
            waveform = torch.mean(waveform, dim=0, keepdim=True)
        return waveform.squeeze().numpy()

    return positive_pairs, negative_pairs, eval_keys, load_fn


# =====================================================================
# EVALUATOR
# =====================================================================
def evaluate_dataset(dataset_name, positive_pairs, negative_pairs, eval_keys,
                      load_fn, ecapa_model, xvect_model):
    print(f"\n--- Running verification on {dataset_name.upper()} ---")

    ecapa_cache, xvect_cache = {}, {}
    for idx, key in enumerate(eval_keys):
        if idx % 100 == 0 or idx == len(eval_keys) - 1:
            print(f"   -> Extracting embeddings: {idx + 1}/{len(eval_keys)}")
        try:
            audio_data = load_fn(key)
            ecapa_emb = get_embedding(audio_data, ecapa_model)
            xvect_emb = get_embedding(audio_data, xvect_model)
        except Exception as e:
            print(f"      [Warning] Skipping {key}: {e}")
            continue
        # Only cache if BOTH succeeded -- keeps the two caches in lockstep
        ecapa_cache[key] = ecapa_emb
        xvect_cache[key] = xvect_emb

    cos = torch.nn.CosineSimilarity(dim=-1)
    results = {}
    for model_name, cache in [("ECAPA-TDNN", ecapa_cache), ("X-Vector", xvect_cache)]:
        pos_scores = [cos(cache[a], cache[b]).item()
                      for a, b in positive_pairs if a in cache and b in cache]
        neg_scores = [cos(cache[a], cache[b]).item()
                      for a, b in negative_pairs if a in cache and b in cache]

        avg_pos = float(np.mean(pos_scores)) if pos_scores else 0.0
        avg_neg = float(np.mean(neg_scores)) if neg_scores else 0.0
        eer, thresh = calculate_eer(pos_scores, neg_scores)

        results[model_name] = {
            "avg_pos": avg_pos, "avg_neg": avg_neg,
            "margin": avg_pos - avg_neg, "thresh": thresh, "eer": eer,
            "n_pos": len(pos_scores), "n_neg": len(neg_scores),
        }
    return results


# =====================================================================
# MAIN
# =====================================================================
def main(get_files_fn):
    print("1. Loading models onto device...")
    ecapa_model = EncoderClassifier.from_hparams(
        source=ECAPA_SOURCE, savedir="tmpdir_ecapa", run_opts={"device": str(device)}
    )
    xvect_model = EncoderClassifier.from_hparams(
        source=XVECT_SOURCE, savedir="tmpdir_xvect", run_opts={"device": str(device)}
    )

    sinhala_pos, sinhala_neg, sinhala_eval, sinhala_load = get_sinhala_trials(
        get_files_fn, NUM_SPEAKERS_SINHALA, UTTS_PER_SPEAKER, RANDOM_SEED
    )
    english_pos, english_neg, english_eval, english_load = get_librispeech_trials(
        NUM_SPEAKERS_SINHALA, UTTS_PER_SPEAKER, RANDOM_SEED
    )

    sinhala_results = evaluate_dataset(
        "Sinhala", sinhala_pos, sinhala_neg, sinhala_eval, sinhala_load,
        ecapa_model, xvect_model,
    )
    english_results = evaluate_dataset(
        "English (LibriSpeech)", english_pos, english_neg, english_eval, english_load,
        ecapa_model, xvect_model,
    )

    print("\n" + "=" * 90)
    print("FINAL COMPARISON: ECAPA-TDNN vs X-VECTOR  (scaled, speaker-diversity aware)")
    print("=" * 90)

    rows = ["Average Same-Speaker Sim", "Average Diff-Speaker Sim",
            "Separation Margin (up=better)", "Decision Threshold",
            "Equal Error Rate (% down=better)", "N positive / N negative trials"]

    def col(res, model):
        r = res[model]
        return [
            f"{r['avg_pos']:.4f}", f"{r['avg_neg']:.4f}", f"{r['margin']:.4f}",
            f"{r['thresh']:.4f}", f"{r['eer']:.2f}%", f"{r['n_pos']} / {r['n_neg']}",
        ]

    summary = pd.DataFrame({
        "Metric": rows,
        "Sinhala: X-Vector": col(sinhala_results, "X-Vector"),
        "Sinhala: ECAPA": col(sinhala_results, "ECAPA-TDNN"),
        "English: X-Vector": col(english_results, "X-Vector"),
        "English: ECAPA": col(english_results, "ECAPA-TDNN"),
    })
    print(summary.to_string(index=False))
    print("=" * 90)
    return summary


if __name__ == "__main__":
    # Plug in your own get_files() here -- must return (local_file_ids, paths_dict)
    try:
        get_files  # noqa: F821 -- defined earlier in your notebook
    except NameError:
        raise RuntimeError(
            "get_files() is not defined. Define it (as in your original notebook) "
            "before running main(get_files)."
        )
    main(get_files)
```
