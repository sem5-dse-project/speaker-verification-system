## Shape check
Dummy 4s @ 16 kHz → Mel / inverted-Mel / LFCC tensors.

```python
wav = torch.randn(16000 * 4)
mel = LogMelSpectrogram()(wav)
imel = InvertedLogMelSpectrogram()(wav)
lfcc = LogLFCC()(wav)
print("mel ", tuple(mel.shape))
print("imel", tuple(imel.shape))
print("lfcc", tuple(lfcc.shape))
```
