# Train Mel / inverted-Mel / LFCC

Same mixed **2017 + PA2019** data as the inverted-Mel mixed experiment. Uses GPU + AMP when CUDA is available.

Set `SMOKE = True` for a short subset run first.

```python
from pathlib import Path
import sys

ROOT = Path.cwd()
if not (ROOT / "experiment_lib.py").exists():
    ROOT = Path(r"D:\speaker-verification-system\replay-cnn-baseline\experiments\lfcc_vs_mel_compare")
sys.path.insert(0, str(ROOT))

import torch
from experiment_lib import FEATURE_TYPES, DEFAULT_ASV17, DEFAULT_PA, train_feature

print("cuda:", torch.cuda.is_available())
if torch.cuda.is_available():
    print(torch.cuda.get_device_name(0))
print("ASV17:", DEFAULT_ASV17)
print("PA:   ", DEFAULT_PA)
```

```python
# --- knobs ---
SMOKE = False          # True → small subset, 2 epochs
FEATURE_LIST = list(FEATURE_TYPES)  # or e.g. ["lfcc"] only
EPOCHS = 2 if SMOKE else 15
PATIENCE = 2 if SMOKE else 4
BATCH_SIZE = 8
MAX_TRAIN = 4000 if SMOKE else 0
MAX_VAL = 1000 if SMOKE else 0
FORCE_CPU = False      # leave False to use NVIDIA GPU
```

## Train loop
Checkpoints land in `runs/{feature}/best_{feature}_mixed_2017_pa2019.pt`.

```python
summaries = []
for ft in FEATURE_LIST:
    print("\n" + "=" * 60)
    print("TRAINING", ft)
    print("=" * 60)
    summaries.append(
        train_feature(
            ft,
            epochs=EPOCHS,
            patience=PATIENCE,
            batch_size=BATCH_SIZE,
            max_train=MAX_TRAIN,
            max_val=MAX_VAL,
            force_cpu=FORCE_CPU,
        )
    )

for s in summaries:
    print(f"{s['feature_type']:14s} mixed-val EER={s['best_val_eer_percent']:.2f}%  -> {s['checkpoint']}")
```
