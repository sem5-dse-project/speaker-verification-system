# Eval LFCC-LA on ASVspoof 2019 LA

Score the trained LFCC checkpoint on LA **dev** (set `SPLIT = "eval"` for official eval).
Compare oracle EER to zero-shot mixed inverted-Mel (~41%).

```python
from pathlib import Path
import sys

from IPython.display import Markdown, display

ROOT = Path.cwd()
if not (ROOT / "train_lib.py").exists():
    ROOT = Path(r"D:\speaker-verification-system\replay-cnn-baseline\experiments\lfcc_la2019")
sys.path.insert(0, str(ROOT))

from train_lib import DEFAULT_CKPT, RUNS_DIR, eval_lfcc_la

print("ckpt exists:", DEFAULT_CKPT.exists(), DEFAULT_CKPT)
```

```python
SPLIT = "dev"       # or "eval"
MAX_UTTS = 0        # 0 = all; e.g. 2000 for smoke
BATCH_SIZE = 8
FORCE_CPU = False
```

```python
out = RUNS_DIR / "eval" / SPLIT
res = eval_lfcc_la(
    DEFAULT_CKPT,
    split=SPLIT,
    batch_size=BATCH_SIZE,
    max_utts=MAX_UTTS,
    force_cpu=FORCE_CPU,
    output_dir=out,
)

md = f"""
| Model | LA {SPLIT} oracle EER |
|-------|------------------------|
| Zero-shot mixed inverted-Mel | ~{res['zero_shot_mixed_imel_la_dev_eer_percent']:.1f}% (dev) |
| **LFCC CNN trained on LA** | **{res['oracle_eer_percent']:.2f}%** |

N={res['n']} (bona={res['n_bonafide']}, spoof={res['n_spoof']})  
Acc @ train thr={res['metrics_at_train_threshold']['accuracy']*100:.1f}%
"""
display(Markdown(md))
print(md)
print("Wrote", out)
```
