# Train LFCC CNN on ASVspoof 2019 LA

Same CNN as the feature-compare experiments; **LFCC** front-end; train on LA only.
Baseline to beat: zero-shot mixed inverted-Mel on LA ≈ **41% EER**.

```python
from pathlib import Path
import sys

ROOT = Path.cwd()
if not (ROOT / "train_lib.py").exists():
    ROOT = Path(r"D:\speaker-verification-system\replay-cnn-baseline\experiments\lfcc_la2019")
sys.path.insert(0, str(ROOT))

import torch
from train_lib import DEFAULT_LA, train_lfcc_la

print("cuda:", torch.cuda.is_available())
if torch.cuda.is_available():
    print(torch.cuda.get_device_name(0))
print("LA:", DEFAULT_LA, "exists", DEFAULT_LA.exists())
```

## Knobs
`SMOKE = True` → small subset, 2 epochs. Then set `False` for full train.

```python
SMOKE = False
EPOCHS = 2 if SMOKE else 15
PATIENCE = 2 if SMOKE else 4
BATCH_SIZE = 8
MAX_TRAIN = 4000 if SMOKE else 0
MAX_VAL = 1000 if SMOKE else 0
FORCE_CPU = False
```

## Train
Checkpoint → `runs/lfcc_la/best_lfcc_la2019.pt`

```python
summary = train_lfcc_la(
    epochs=EPOCHS,
    patience=PATIENCE,
    batch_size=BATCH_SIZE,
    max_train=MAX_TRAIN,
    max_val=MAX_VAL,
    force_cpu=FORCE_CPU,
)
print(
    f"best val EER={summary['best_val_eer_percent']:.2f}%"
    f"  (zero-shot baseline ~{summary['zero_shot_baseline_la_dev_eer_percent']:.1f}%)"
)
print("ckpt:", summary["checkpoint"])
```
